### Dataset description findings

- There is < 1% of rows containing duplicate image_url
- We only check if there are duplicate image_url and not other features such as filename, or amazon_index. Since there could be multiple indices for the same image_url
- There is a one-to-many relationship between the image_url and category_id. (One image_url can have more than one category_id).

## Steps done for Duplicate rows
1. Keep those rows where `image_url` is the same but `category_id` is different
2. Else, if `image_url` is same, we will keep the first instance and drop the others

- We will be keeping those rows with same `image_url` but different `category_id` regardless of other features, since there could be relationship between those multiple categories and an image
- We just drop the duplicated rows, since it is less than 1% of the total

```python
train_duplicated_rows_to_keep = train_df.drop_duplicates(subset=['image_url', 'category_id'])
val_duplicated_rows_to_keep = val_df.drop_duplicates(subset=['image_url', 'category_id'])
test_duplicated_rows_to_keep = test_df.drop_duplicates(subset=['image_url', 'category_id'])
```

```python
print("Before vs. After dropping of duplicates")
print(f"Train: {len(train_df)} vs. {len(train_duplicated_rows_to_keep)}")
print(f"Val: {len(val_df)} vs. {len(val_duplicated_rows_to_keep)}")
print(f"Test: {len(test_df)} vs. {len(test_duplicated_rows_to_keep)}")
```

## Image color
 The paper by Iwana et al (https://arxiv.org/pdf/1610.09204), describes features that their CNN was able to learn from the dataset. One of the high-level features was the association of colors to certain genres of books

Majority of categories dont have any outliers in color except a few:
- Category ID 18, 22, and 24 have high values for blue color
- Category ID 24 have high values for all colors
- Category ID 11, 17 has low values for all colors
Some genres have outliers for the average color of their book covers, but majority seem to not

The paper by Iwana et al describes certain colors such as 'beige', 'tan', 'green' , 'yellow' that are associated with some or multiple genres.
The RGB color channels individually only describe certain colors, but a mixture of multiple colors could, although not visible with just the 3 color channel heatmap

```python
fig, axes = plt.subplots(3, 1, figsize=(20, 15))
fig.suptitle("Red,Green, Blue Color Distribution by Genry", fontsize=16)
sns.boxplot(x="category_id", y="red", data=image_df, palette=['r'], ax=axes[0])
sns.boxplot(x="category_id", y="green", data=image_df, palette=['g'], ax=axes[1])
sns.boxplot(x="category_id", y="blue", data=image_df, palette=['b'], ax=axes[2])
plt.show()
```
