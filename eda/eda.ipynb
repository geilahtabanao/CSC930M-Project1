## Steps done for Duplicate rows
1. Keep those rows where `image_url` is the same but `category_id` is different
2. Else, if `image_url` is same, drop those rows since count of duplicate is < 1%

```python
train_duplicated_rows_to_keep = train_df.drop_duplicates(subset=['image_url', 'category_id'])
val_duplicated_rows_to_keep = val_df.drop_duplicates(subset=['image_url', 'category_id'])
test_duplicated_rows_to_keep = test_df.drop_duplicates(subset=['image_url', 'category_id'])
```

```python
print("Before vs. After dropping of duplicates")
print(f"Train: {len(train_df)} vs. {len(train_duplicated_rows_to_keep)}")
print(f"Train: {len(val_df)} vs. {len(val_duplicated_rows_to_keep)}")
print(f"Train: {len(test_df)} vs. {len(test_duplicated_rows_to_keep)}")
```
