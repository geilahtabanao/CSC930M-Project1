```python
import pandas as pd

from sklearn.model_selection import train_test_split
```

```python
X_df = pd.read_csv(r"..\dataset_updated\book30-listing-train.csv", 
                   names=["amazon_index_asin", "filename", "image_url", "title", "author", "category_id", "category"],
                   index_col="filename")
```

```python
X_df.head()
```

```python
X_df.describe()
```

```python
X_df.info()
```

```python
y_df = pd.read_csv(r"..\dataset_updated\bookcover30-labels-train.txt",
                   sep=" ",
                   names=["filename", "category_id"],
                   index_col="filename")
```

```python
y_df.head()
```

```python
y_df.describe()
```

```python
y_df.info()
```

```python
X_train, X_val, y_train, y_val = train_test_split(X_df, y_df, test_size=(1/9), random_state=42, stratify=y_df)
```

```python
# Check if the split is stratified

y_df.value_counts(normalize="True")
```

```python
y_train.value_counts(normalize="True")
```

```python
y_val.value_counts(normalize="True")
```

```python
train_folder_path = r"..\dataset_updated\train"
val_folder_path = r"..\dataset_updated\validation"
test_folder_path = r"..\dataset_updated\test"
```

```python
X_train.to_csv(train_folder_path+"/listings_train.csv")
y_train.to_csv(train_folder_path+"/labels_train.csv")
```

```python
X_val.to_csv(val_folder_path+"/listings_val.csv")
y_val.to_csv(val_folder_path+"/labels_val.csv")
```
